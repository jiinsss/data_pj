```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
spark = (
    SparkSession.builder
    .appName("dd")
    .master("local[*]") 
    .config("spark.driver.memory", "4g")
    .getOrCreate()
)
```

```python
import os
BASE_DIR = os.getenv("DATA_DIR", "../")

df_oct = spark.read.csv(f"{BASE_DIR}/raw/2019-Oct.csv", header=True, inferSchema=True)
df_nov = spark.read.csv(f"{BASE_DIR}/raw/2019-Nov.csv", header=True, inferSchema=True)

df_oct.write.mode("overwrite").parquet(f"{BASE_DIR}/processed/2019-10/")
df_nov.write.mode("overwrite").parquet(f"{BASE_DIR}/processed/2019-11/")
```
